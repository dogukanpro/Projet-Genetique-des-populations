```python
ligne_pdb = "ATOM      1  CA  ALA A   1      12.268   5.210  41.725  1.00 24.83           C"

if ligne_pdb.startswith("ATOM"):
    #Slicing
    nom_atome = ligne_pdb[12:16].strip()
    residu    = ligne_pdb[17:20].strip()
    coord_x   = float(ligne_pdb[30:38])
    coord_y   = float(ligne_pdb[38:46])
    coord_z   = float(ligne_pdb[46:54])
    print(f"Atome : {nom_atome} ({residu})")
    print(f"Coordonnées : X={coord_x}, Y={coord_y}, Z={coord_z}")
```
