Sujet : filtrage d'atome par type (par exemple ici pour CA), grace a une class

```python
class Atome:
    def __init__(self, nom, x, y, z):
        self.nom = nom
        self.x = x
        self.y = y
        self.z = z

# Simulation d'un extrait de fichier PDB
lignes_pdb = [
    "ATOM      1  N   ALA A   1      12.268   5.210  41.725",
    "ATOM      2  CA  ALA A   1      12.800   6.100  42.500", # On veut celui-là
    "ATOM      3  C   ALA A   1      13.500   5.800  43.200",
    "ATOM      4  N   GLY A   2      14.100   6.500  44.100",
    "ATOM      5  CA  GLY A   2      14.800   7.200  45.000"  # Et celui-là
]

squelette = []

# Parcours et filtrage
for ligne in lignes_pdb:
    nom = ligne[12:16].strip()
    if nom == "CA":
        x = float(ligne[30:38])
        y = float(ligne[38:46])
        z = float(ligne[46:54])
        # On crée l'objet et on l'ajoute à notre liste
        squelette.append(Atome(nom, x, y, z))

print(f"Filtrage terminé. Atomes dans le squelette : {len(squelette)}")
for a in squelette:
    print(f"{a.nom} trouvé en position ({a.x}, {a.y}, {a.z})")
```
