```python
#Définition d'une class
class Atome:
    def __init__(self, nom, x, y, z, masse):
        self.nom = nom
        self.x = x
        self.y = y
        self.z = z
        self.masse = masse

#2 exemple d'objets
carbone_alpha = Atome("CA", 12.5, 4.8, -0.1, 12.01)
oxygene = Atome("O", 13.2, 5.1, 0.5, 15.99)

#Instanciation des données
print(f"Atome : {carbone_alpha.nom}")
print(f"Position Z : {carbone_alpha.z} Å")
print(f"Masse : {carbone_alpha.masse} g/mol")

#Calcul entre objets
dist_x = oxygene.x - carbone_alpha.x
print(f": {round(dist_x, 2)} Å")
```
