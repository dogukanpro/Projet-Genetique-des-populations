Sujet : module re avec fonctiopn dindalle

```python
import re

sequence = "ATGCAGGCATACGGTTAGG"

motif = "A.GG"

#Il va chercher le motif donc le A puis une autre lettre de la sequence puis 2 G
resultats = re.findall(motif, sequence)

print(f"Séquence : {sequence}")
print(f"Motifs trouvés : {resultats}")
```
