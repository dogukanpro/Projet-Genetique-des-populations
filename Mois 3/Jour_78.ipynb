Sujet : Fusion de deux listes avec union () et intersection () pour voir s'il ya des élément commun

```python
pop_occitanie = {"Allèle_1", "Allèle_2", "Allèle_3"}
pop_provence = {"Allèle_3", "Allèle_4", "Allèle_5"}

pool_global = pop_occitanie.union(pop_provence)

print(f"Variants uniques après migration : {pool_global}")

# Intersection pour voir s'il ya un élément commun aux deux listes
points_communs = pop_occitanie.intersection(pop_provence)
print(f"Allèles partagés par les deux groupes : {points_communs}")
```
