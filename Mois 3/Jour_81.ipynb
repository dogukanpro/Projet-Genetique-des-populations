Sujet : les dates avec datetime, timedelta...(par exemple pour calculer les dates des divergence génétique)

```python
from datetime import datetime, timedelta

aujourdhui = datetime.now()
print(f"Date du jour de l'analyse : {aujourdhui.strftime('%d/%m/%Y')}")

jours_passés = timedelta(days=36500)

date_divergence = aujourdhui - jours_passés

print(f"La divergence a eu lieu en l'an : {date_divergence.strftime('%Y')}")
```
