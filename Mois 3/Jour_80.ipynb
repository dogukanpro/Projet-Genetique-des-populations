Sujet : random choice, pour simuler le hasard

```python
import random

population_initiale = ["A", "A", "A", "A", "B", "B", "B", "B"]
print(f"Génération 0 : {population_initiale}")


generation_suivante = []

#choix va prendre au hasard une valeur de la population initiale, et après on va l'ajouter à la génération suivante
for i in range(8):
    choix = random.choice(population_initiale)
    generation_suivante.append(choix)

print(f"Génération 1 : {generation_suivante}")


count_A = generation_suivante.count("A")
count_B = generation_suivante.count("B")
print(f"Score : A={count_A}, B={count_B}")
```
