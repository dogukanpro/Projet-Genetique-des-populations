Sujet : module re, méthode group pour recuperer un extrait précis

```python
import re


brut = "ZONE_INUTILE_DEBUT_ATGCGG_FIN_ZONE_INUTILE"

#On capture tout ce qu'il y a entre DEBUT_ ET _FIN
motif = "DEBUT_(.*)_FIN"

recherche = re.search(motif, brut)

if recherche:
    gene_isole = recherche.group(1)
    print(f"Gène extrait avec succès : {gene_isole}")
```
