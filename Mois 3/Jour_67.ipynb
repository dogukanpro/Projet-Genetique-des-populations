Sujet : fonctions split et float, pour faire des calculs (ici pour calculer une moyenne)

```python
total_gc = 0
compteur = 0

with open("resultats_occitan.csv", "r") as f:
    next(f)
    
    for ligne in f:
        colonnes = ligne.strip().split(",")
        valeur_gc = float(colonnes[1])
        
        total_gc += valeur_gc
        compteur += 1

moyenne = total_gc / compteur

print(f"Analyse de {compteur} échantillons terminée.")
print(f"Moyenne du taux de GC : {moyenne:.2f}%")
```
