Sujet : Vérification avec isinstance()

```python
donnees_brutes = ["Modern_Occitan_01", 42.5, "Old_Occitan_01", 48.2, "Analyse_Error", 0.0]

taux_gc_extraits = []

#Je regarde si dans donnees brutes il y a des donnees de type float et les ajoute à la liste
for donnee in donnees_brutes:
    if isinstance(donnee, float):
        taux_gc_extraits.append(donnee)

#Affichage
print(f"Le taux de gc : {taux_gc_extraits}")
print(f"La moyenne en pourcantage : {sum(taux_gc_extraits)/len(taux_gc_extraits):.2f}%")
```
