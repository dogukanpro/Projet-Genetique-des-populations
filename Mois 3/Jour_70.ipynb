Sujet : Tri avec sort ()

```python
echantillons = ["Old_Occitan_01", "Modern_Occitan_01", "Old_Celt_01", "Modern_Celt_01"]

echantillons.sort()
print(f"Tri alphabétique (de base) : {echantillons}")

#Tri par rapport à la taille
echantillons.sort(key=len)
print(f"Tri par longueur : {echantillons}")

#Tri par rapport à la taille mais en sens inversé donc du plus long au plus court
echantillons.sort(key=len, reverse=True)
print(f"Tri sens inversé : {echantillons}")
```
