Sujet : suite module re avec la fonction "^" pour voir si quelquechose commence par, et "$" pour voir si quelquechose finit par

```python
import re

sequence = "ATGCAGGCATACGGTTAAAA"
motif_complet = "^ATG.*AAAA$"

if re.search(motif_complet, sequence):
    print("Séquence trouvée")
else:
    print("Séquence non trouvée")

#Pour rechercher spécifiquement si sa finit par AAAAA
if re.search("AAAA$", sequence):
    print("🧬 Présence d'une queue poly-A détectée.")
```
