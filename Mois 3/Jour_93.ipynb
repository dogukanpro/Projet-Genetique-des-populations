Sujet : module math, coordonnées 3D

```python
import math


#J'utilise un utilise un tuple car ces coordonnées sont des données fixes
atome_C = (1.25, 0.48, -0.10)


print(f"📍 Position de l'atome de Carbone :")
print(f"X: {atome_C[0]}, Y: {atome_C[1]}, Z: {atome_C[2]}")

#Formule : racine(x² + y² + z²)
distance_origine = math.sqrt(atome_C[0]**2 + atome_C[1]**2 + atome_C[2]**2)

print(f"📏 Distance du centre : {round(distance_origine, 2)} Å")
```
