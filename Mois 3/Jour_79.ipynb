Sujet : La fonction fronzenset qui va nous permettre de créer une liste non modifiable par la suite

```python
variants_liste = {"Occ_01", "Occ_02", "OcC_03"}

reference_occitanie = frozenset(variants_liste)

print(f"Référence sécurisée : {reference_occitanie}")

#Tentative de modification, pour montrer qu'on ne peut pas modifier
try:
    reference_occitanie.add("Occ_04") 
except AttributeError:
    print("Action interdite : On ne peut pas modifier un frozenset !")
```
