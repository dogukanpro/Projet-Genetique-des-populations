Sujet : Utilisation du module sys

```python
import sys

#Je crée une séquence "chromosome" (répétition de 1 million de bases)

chromosome_1 = "ATGC" * 250000 

#Mesure la taille de cet objet en mémoire (en octets)
taille_octets = sys.getsizeof(chromosome_1)

#Conversion en kilo octets
taille_kb = taille_octets / 1024

print(f"Longueur de la séquence est de : {len(chromosome_1)} nucléotides")
print(f"La taille est de : {taille_kb:.2f} KB")

#Exemple pour supprimer, pour libérer de la place une fois utilisé
del chromosome_1
```
