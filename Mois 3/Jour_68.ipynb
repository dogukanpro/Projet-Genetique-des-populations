```python
echantillons_riches = []
seuil = 45.0

with open("resultats_occitan.csv", "r") as f:
    next(f)
    
    for ligne in f:
        colonnes = ligne.strip().split(",")
        nom = colonnes[0]
        valeur_gc = float(colonnes[1])
        
        if valeur_gc > seuil:
            echantillons_riches.append(nom)

print(f"Analyse terminée. Seuil : > {seuil}%")
print(f"{len(echantillons_riches)} échantillons correspondent au critère :")
for nom in echantillons_riches:
    print(f" - {nom}")
```
