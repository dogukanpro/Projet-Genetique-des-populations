Sujet : la fontion yield

```python

def lire_genome_ancien():
    fragments = ["ATGC...", "TTAA...", "GGC..."] 
    
    for f in fragments:
        print("Extraction d'un fragment")
        yield f


genome_gen = lire_genome_ancien()

#Affichage
print("Début de l'analyse sans charger tout le fichier :")
for fragment in genome_gen:
    print(f"Analyse du fragment : {fragment}")
```
