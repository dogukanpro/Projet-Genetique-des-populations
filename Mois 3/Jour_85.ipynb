Sujet : suite re, les accolades pour donner des intervalles

```python
import re

sequence = "ATGGGCATGGGGGCTAGGGGGGGT"

#motif entre 3 et 5 fois G
motif_repetition = "G{3,5}"

resultats = re.findall(motif_repetition, sequence)

print(f"Séquence analysée : {sequence}")
print(f"Zones de répétition (3 à 5 G) : {resultats}")
```
