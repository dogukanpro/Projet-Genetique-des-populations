Sujet : la comprehension de liste pour optimiser et aller plus vite comparé à une boucle for et if classique

```python
taux_bruts = [42.1, 48.5, 39.0, 51.2, 44.8, 47.3]

#C'est ici que la comprehnsion de liste se met en place, j'ai écrit garde t pour chaque t dans taux bruts seulement si t > 45
taux_optimises = [t for t in taux_bruts if t > 45]

print(f"Analyse rapide terminée : {taux_optimises}")
```
