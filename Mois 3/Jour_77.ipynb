Sujet : Les tuple, des données non modifiable, les "()" vont protéger les données en disant a python de ne pas permettre la modification de ce qui est entre parenthèse 

```python
paire_1 = ("A", "T")
paire_2 = ("G", "C")

print(f"Paire 1 {paire_1} et paire 2 {paire_2}")


try:
    paire_1[0] = "G" 
except TypeError:
    print("Erreur bloquée : Un tuple ne peut pas être modifié !")


liste_avec_tuple = [("A", "T"), ("T", "A"), ("G", "C"), ("C", "G")]

for base_a, base_b in liste_avec_tuple:
    print(f"Liaison stable : {base_a} <-> {base_b}")
```
