Sujet : utilisation de args(*), pour pouvoir manipuler plusieurs variable, sans avoir à préciser à chaque fois, comme ici ou je vais faire une méthode pour pouvoir calculer la taille, le taux de GC en pourcentage

```python
def analyse_pool_genetique(*sequences):
    total_bases = 0
    total_gc = 0
    
    for s in sequences:
        total_bases += len(s)
        total_gc += (s.count("G") + s.count("C"))
    
    moyenne_globale = (total_gc / total_bases) * 100
    return moyenne_globale


resultat_2 = analyse_pool_genetique("ATGC", "ATGG")
print(f"Moyenne (2 variables) : {resultat_2:.2f}%")

resultat_4 = analyse_pool_genetique("ATGC", "ATGG", "CCGG", "TTAA")
print(f"Moyenne (4 variables) : {resultat_4:.2f}%")
```
