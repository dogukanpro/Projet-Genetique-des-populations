```python
def calculer_gc(sequence):
    # On nettoie et on compte
    sequence = sequence.upper()
    g = sequence.count('G')
    c = sequence.count('C')
    taux = (g + c) / len(sequence) * 100
    return round(taux, 2) #Round : arrondie le résultat; à 2 chiffres après la virgule

echantillon_1 = "ATGCGGTA"
echantillon_2 = "CCGGTACTAAAGTCG"

resultat_1 = calculer_gc(echantillon_1)
resultat_2 = calculer_gc(echantillon_2)

print(f"Taux échantillon 1 : {resultat_1}%")
print(f"Taux échantillon 2 : {resultat_2}%")
```
