Sujet : Suite module re fonction sub pour remplacer les motifs

```python
import re

sequence_brute = "ATG-55-CGG--12-TTA"
print(f"Brute : {sequence_brute}")


etape_1 = re.sub("-", "", sequence_brute)

sequence_propre = re.sub(r"\d", "", etape_1)

print(f"Propre : {sequence_propre}")


sequence_alerte = re.sub(r"[^ATGC]", "!", sequence_brute)
print(f"Alerte : {sequence_alerte}")
```
