```python
#Je met les attributs en majuscule pour que Python les voient comme des constantes
MASSE_H = 1.008
MASSE_C = 12.011
MASSE_N = 14.007
MASSE_O = 15.999

#Calcul pour une molécule d'eau (H2O)
nb_h = 2
nb_o = 1

masse_eau = (nb_h * MASSE_H) + (nb_o * MASSE_O)

print(f"Molécule : Eau (H2O)")
print(f"Masse moléculaire estimée : {round(masse_eau, 3)} g/mol")


masse_methane = MASSE_C + (4 * MASSE_H)
print(f"Molécule : Méthane (CH4)")
print(f"Masse moléculaire estimée : {round(masse_methane, 3)} g/mol")
```
