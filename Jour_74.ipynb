Sujet : Utilisaton des doubles **, pour pouvoir créer un programme évolutif, c'est à dire qui permet des réglages personnalisées sans avoir à définir avant précisement 

```python
def simuler_mutation(sequence, **reglages):
    print(f"Séquence d'origine : {sequence}")
    
    
    if "taux" in reglages:
        print(f"Application d'un taux de mutation de {reglages['taux']}%")
    
    if "environnement" in reglages:
        print(f"Simulation dans un milieu : {reglages['environnement']}")

    
    return sequence.replace("A", "G") # Une mutation fictive


print("--- Test 1 ---")
simuler_mutation("ATGC", taux=0.05, environnement="Grotte humide")

print("\n--- Test 2 (Paramètres différnts) ---")
simuler_mutation("ATGC", chercheur="01", labo="01")
```
