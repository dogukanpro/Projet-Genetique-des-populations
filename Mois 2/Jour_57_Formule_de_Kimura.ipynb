Sujet : Le module math (fonctions exp, log, sqrt)

```python
import math

avantage_selectif = 0.05
taille_population = 1000

print("Simulation de survie d'un nouvel allèle :")

#Formule de Kimura
probabilite = (1 - math.exp(-2 * avantage_selectif)) / (1 - math.exp(-4 * taille_population * avantage_selectif))

#J'arrondis à deux chiffres après la virgule
pourcentage = round(probabilite * 100, 2)

print(f"Pour un avantage de {avantage_selectif * 100}%")
print(f"La probabilité que la mutation s'installe est de : {pourcentage}%.")
```
