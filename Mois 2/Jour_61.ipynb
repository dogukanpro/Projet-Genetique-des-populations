Sujet : révision du premier mois, calcul de pourcentage

```python
adn = "ATGCGGTACGATACGCTAG"

nb_g = adn.count("G")
nb_c = adn.count("C")


total_gc = nb_g + nb_c
taux_gc = (total_gc / len(adn)) * 100

print(f"Séquence : {adn}")
print(f"Nombre de G+C : {total_gc}")
print(f"📈 Taux de GC : {taux_gc:.2f}%")
```
