Sujet : Accumuler le texte d'un fichier fasta, en sautant les lignes si on rencontre le signe ">" dans le fichier fasta, grace a la boucle for et if à l'intérieur

```python
with open("sequence.fasta", "r") as f:
    sequence_finale = ""
    
    for ligne in f:
        #Nettoyer les espaces et sauts de ligne
        ligne = ligne.strip()
        
        #Si la ligne est un ">", on passe à la ligne suivante
        if ligne.startswith(">"):
            continue
        
        #sinon, on ajoute la ligne à notre séquence
        sequence_finale += ligne

#Affichage
print(f"Séquence extraite : {sequence_finale}")
print(f"Longueur réelle : {len(sequence_finale)} bases")
```
