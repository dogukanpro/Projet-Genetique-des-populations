Sujet : Porgramme qui compte les répétitions dans un fichier 

```python
with open("occitan_genetics.fasta", "w") as f:
    f.write(">Modern_Occitan_01\nATGCGTACGG\n")
    f.write(">Old_Occitan_01\nATGCGTACCC\n")
    f.write(">Modern_Occitan_02\nTTGCGTACAA\n")

nb_sequences = 0

with open("occitan_genetics.fasta", "r") as f:
    for ligne in f:
        if ligne.startswith(">"):
            nb_sequences += 1

print(f"Nombre d'échantillons détectés : {nb_sequences}")
```
