Sujet : Utilisation de breakpoint() pour trouver les erreurs plus facilement, sans avoir à parcourir toutes les lignes de code

```python
sequences = ["ATGC", "ATZC", "CCGG"]

print("Début de l'analyse qualité...")

for s in sequences:
    #Si le programme trouve une caractère qui n'est pas A OU T OU G OU C il execute le print puis le breakpoint
    if any(base not in "ATGC" for base in s):
        print(f"Anomalie détectée dans : {s}")
        breakpoint() 
        
    print(f"Séquence {s} validée.")

print("Analyse terminée.")
```
