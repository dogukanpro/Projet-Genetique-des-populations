Sujet : Calculateur de distance inter-atomique (Pyhtagore)

```python
import math


atome_A = (1.1, 2.5)
atome_B = (2.3, 3.1)

#Calcul de la différence sur chaque axe
dx = atome_B[0] - atome_A[0]
dy = atome_B[1] - atome_A[1]

#Pythagore
distance = math.sqrt(dx**2 + dy**2)

print(f"Atome A : {atome_A}")
print(f"Atome B : {atome_B}")
print(f"Distance calculée : {round(distance, 3)} Å")

#Vérification
if distance < 1.6:
    print("Ces atomes sont probablement liés.")
else:
    print("Ces atomes sont trop loin pour être liés directement.")
```
