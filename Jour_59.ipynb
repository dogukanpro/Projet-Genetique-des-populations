Sujet : révision du 1er mois, nettoyage de chaine de caracteres avec strip, replace et upper

```python

adn_brut = "  atgcGtaCcG \n "

# .strip() retire les espaces et \n au début et à la fin
adn_propre = adn_brut.strip()

# .upper() transforme tout en MAJUSCULES
adn_propre = adn_propre.upper()

# .replace() peut supprimer un caractère au milieu (ex: un tiret)
# adn_propre = adn_propre.replace("-", "")

print(f"Original : '{adn_brut}'")
print(f"Nettoyé  : '{adn_propre}'")
print(f"Longueur : {len(adn_propre)} bases")
```
