Sujet : suite module re, avec la fonction sub pour remplacer, substituer (utile pour corriger très rapidement la meme erreur qui apparait plusieurs fois)

```python
import re

sequence_brute = "ATG-55-CGG--12-TTA"
print(f"Brute : {sequence_brute}")

#Je remplace les tirets par le vide
etape_1 = re.sub("-", "", sequence_brute)

#Je remplace tout ce qui n'est pas A, T, G ou C par le vide (pour enlever les chiffres)
# [^...] pour dire "tout sauf ce qu'il y a entre corchets"
sequence_corrige = re.sub(r"[^ATGC]", "", sequence_brute)
print(f"Première étape: {etape_1}, puis la correction finale : {sequence_corrige}")
```
